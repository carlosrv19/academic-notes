## Chapter 2: Using Surveys Studies to Gather Data

In this chapter we deep dive in some types of research study. They can be **observational** or **experimental** studies. 
- In observational studies the researcher is limited only to record information from subjects (or units). No intervention or manipulation of the process involved in the data generation.
- In contrast, in experimental studies the researcher purposely manipulates some variables associated with the study. Often, this manipulation is the result of randomly assigning a predefined *treatment* to the experimental unit. For example, in a clinical trial, the researcher randomly assigns a patient to one of many treatments. In this case, the variable being manipulated is "treatment" that takes values (levels) `treatment_A`, `treatment_B`, etc. Experimental studies are executed with maximum control so that if a difference is found between different groups, that difference is attributed only to the fact that groups have different treatments. For that reason, in experimental studies, one group is very similar in characteristics to other group(s), *except* in that they have been given a different treatment.
